```python
# Script to download the h5 pfss files for a particular month or day
# 
# These are the DeRosa ones that are part of the sswidl pfss package
# https://hesperia.gsfc.nasa.gov/ssw/packages/pfss/
# https://www.lmsal.com/~derosa/pfsspack/
# 
# Files urls are like:
# https://www.lmsal.com/solarsoft/archive/ssw/pfss_links_v2/Bfield_20030607_120400.h5
# 
# 6 Oct 2026 IGH
```

```python
import calendar
import subprocess
from pathlib import Path

BASE_URL = "https://www.lmsal.com/solarsoft/archive/ssw/pfss_links_v2"

#  Get the midday pfss for a whole month
def download_pfss_month(year, month, outdir="pfss_org"):
    """Download the daily Bfield_YYYYMMDD_120400.h5 files for a month with wget.

    Files are saved in outdir/<year>/. Existing files are skipped, and days
    that don't exist on the server are ignored. Returns the list of paths.
    """
    yeardir = Path(outdir) / f"{year}"
    yeardir.mkdir(parents=True, exist_ok=True)
    paths = []
    days_in_month = calendar.monthrange(year, month)[1]
    for day in range(1, days_in_month + 1):
        name = f"Bfield_{year}{month:02d}{day:02d}_120400.h5"
        path = yeardir / name
        if path.exists():
            print(f"[{day:02d}/{days_in_month}] Already exists, skipping: {name}", flush=True)
            paths.append(path)
            continue

        print(f"[{day:02d}/{days_in_month}] Downloading: {name}", flush=True)
        result = subprocess.run(
            ["wget", "-q", "-O", str(path), f"{BASE_URL}/{name}"]
        )
        if result.returncode != 0:
            path.unlink(missing_ok=True)  # Remove any partial file left by wget.
            print(f"[{day:02d}/{days_in_month}] Not available: {name}", flush=True)
            continue
        print(f"[{day:02d}/{days_in_month}] Finished: {name}", flush=True)
        paths.append(path)

    print(f"Downloaded or found {len(paths)} of {days_in_month} files in {yeardir}")
    return paths

# Get the midday pfss for a specific day
def download_pfss_day(year, month, day,outdir="pfss_org"):
    """Download the daily Bfield_YYYYMMDD_120400.h5 files for a specific day with wget.

    Files are saved in outdir/<year>/. Existing files are skipped, and days
    that don't exist on the server are ignored. Returns the list of paths.
    """
    yeardir = Path(outdir) / f"{year}"
    yeardir.mkdir(parents=True, exist_ok=True)
    paths = []
    name = f"Bfield_{year}{month:02d}{day:02d}_120400.h5"
    path = yeardir / name
    if path.exists():
        print(f"Already exists, skipping: {name}", flush=True)
        paths.append(path)
        return "Already exists"
    
    print(f"Downloading: {name}", flush=True)
    result = subprocess.run(
        ["wget", "-q", "-O", str(path), f"{BASE_URL}/{name}"]
    )
    if result.returncode != 0:
        path.unlink(missing_ok=True)  # Remove any partial file left by wget.
        print(f"Not available: {name}", flush=True)
        return "Not available"
    print(f"Finished: {name}", flush=True)
    paths.append(path)

    return paths
```

```python
# Ones from the Loumou 2018 paper Fig 8
path=download_pfss_day(2003,6,7)
path=download_pfss_day(2006,6,8)
path=download_pfss_day(2011,6,16)
path=download_pfss_day(2012,6,6)
path=download_pfss_day(2014,6,10)
path=download_pfss_day(2015,6,6)
```
